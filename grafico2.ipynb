```python
#implementacion del algoritmo de la biseccion
import numpy as np
import matplotlib.pyplot as plt

# Definir la función
def f(x):
    return (1/4) * (x**3 + 3*x**2 - 6*x - 8)

# Implementar el método de bisección
def metodo_biseccion(f, a, b, TOL, N0):
    i = 1
    FA = f(a)
    while i <= N0:
        p = a + (b - a) / 2  # Calcular pᵢ
        FP = f(p)
        if FP == 0 or (b - a) / 2 < TOL:  # Verificar condición de parada
            return p  # Solución encontrada
        i += 1
        if FA * FP > 0:  # Actualizar los extremos del intervalo
            a = p
            FA = FP
        else:
            b = p
    return None  # Si no se encuentra solución

# Encontrar todas las raíces en un intervalo dividiendo en subintervalos
def encontrar_raices(f, intervalo, pasos, TOL, N0):
    raices = []
    x_vals = np.linspace(intervalo[0], intervalo[1], pasos)
    for i in range(len(x_vals) - 1):
        a, b = x_vals[i], x_vals[i + 1]
        if f(a) * f(b) < 0:  # Verificar cambio de signo
            raiz = metodo_biseccion(f, a, b, TOL, N0)
            if raiz is not None:
                raices.append(raiz)
    return raices

# Parámetros
intervalo = [-5, 5]
pasos = 100
TOL = 1e-5
N0 = 100

# Encontrar raíces
raices = encontrar_raices(f, intervalo, pasos, TOL, N0)

# Crear el gráfico
x = np.linspace(intervalo[0], intervalo[1], 1000)
y = f(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='f(x) = (1/4)(x^3 + 3x^2 - 6x - 8)', color='blue')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # Eje x
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')  # Eje y

# Resaltar las raíces en el gráfico
for raiz in raices:
    plt.scatter(raiz, 0, color='red', zorder=5, label=f'Raíz: x = {raiz:.5f}')

# Agregar etiquetas y título
plt.title('Gráfico de f(x) con raíces resaltadas')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()

# Mostrar el gráfico
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Definir la función
def f(x):
    return (1/4) * (x**3 + 3*x**2 - 6*x - 8)

# Implementar el método de bisección con conteo de iteraciones
def metodo_biseccion(f, a, b, TOL, N0):
    i = 1
    FA = f(a)
    while i <= N0:
        p = a + (b - a) / 2  # Calcular pᵢ
        FP = f(p)
        if FP == 0 or (b - a) / 2 < TOL:  # Verificar condición de parada
            return p, i  # Retornar la raíz y el número de iteraciones
        i += 1
        if FA * FP > 0:  # Actualizar los extremos del intervalo
            a = p
            FA = FP
        else:
            b = p
    return None, i  # Si no se encuentra solución, retornar el número de iteraciones

# Encontrar todas las raíces y contar iteraciones
def encontrar_raices_con_iteraciones(f, intervalo, pasos, TOL, N0):
    raices = []
    iteraciones = []
    x_vals = np.linspace(intervalo[0], intervalo[1], pasos)
    for i in range(len(x_vals) - 1):
        a, b = x_vals[i], x_vals[i + 1]
        if f(a) * f(b) < 0:  # Verificar cambio de signo
            raiz, iters = metodo_biseccion(f, a, b, TOL, N0)
            if raiz is not None:
                raices.append(raiz)
                iteraciones.append(iters)
    return raices, iteraciones

# Parámetros
intervalo = [-5, 5]
pasos = 100
TOL = 1e-5
N0 = 100

# Encontrar raíces y sus iteraciones
raices, iteraciones = encontrar_raices_con_iteraciones(f, intervalo, pasos, TOL, N0)

# Imprimir las soluciones y el número de iteraciones
print("Soluciones encontradas y número de iteraciones:")
for i, (raiz, iters) in enumerate(zip(raices, iteraciones), start=1):
    print(f"Raíz {i}: x = {raiz:.5f}, Iteraciones = {iters}")

# Crear un rango de valores para x
x = np.linspace(intervalo[0], intervalo[1], 1000)

# Calcular los valores de f(x)
y = f(x)

# Crear el gráfico
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='f(x) = (1/4)(x^3 + 3x^2 - 6x - 8)', color='blue')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # Eje x
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')  # Eje y

# Resaltar las raíces en el gráfico
for raiz in raices:
    plt.scatter(raiz, 0, color='red', zorder=5, label=f'Raíz: x = {raiz:.5f}')

# Agregar etiquetas y título
plt.title('Gráfico de f(x) con raíces resaltadas')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()

# Mostrar el gráfico
plt.show()

# Crear una tabla con las raíces y sus iteraciones
tabla = pd.DataFrame({
    'Raíz': [f"x = {raiz:.5f}" for raiz in raices],
    'Iteraciones': iteraciones
})

# Mostrar la tabla
print("\nTabla de raíces e iteraciones:")
print(tabla)
```
